```python
import pandas as pd
import random
import faker

fake = faker.Faker()

# Number of records
NUM_CUSTOMERS = 1000
NUM_SELLERS = 100
NUM_CATEGORIES = 10
NUM_PRODUCTS = 500
NUM_ORDERS = 2000
NUM_ORDER_ITEMS = 5000
NUM_PAYMENTS = 2000
NUM_SHIPMENTS = 2000
NUM_INVENTORY = 500

# Sample categories & products from Amazon
categories = [
    "Electronics", "Clothing", "Books", "Home & Kitchen", "Beauty & Personal Care",
    "Sports & Outdoors", "Toys & Games", "Automotive", "Grocery", "Health & Household"
]

products = {
    "Electronics": ["Smartphone", "Laptop", "Headphones", "Smartwatch", "Bluetooth Speaker"],
    "Clothing": ["Jeans", "T-Shirt", "Jacket", "Sneakers", "Sweater"],
    "Books": ["Fiction Novel", "Self-Help Book", "Cookbook", "Science Textbook", "Mystery Thriller"],
    "Home & Kitchen": ["Blender", "Coffee Maker", "Air Fryer", "Vacuum Cleaner", "Toaster"],
    "Beauty & Personal Care": ["Shampoo", "Lipstick", "Face Cream", "Perfume", "Nail Polish"],
    "Sports & Outdoors": ["Yoga Mat", "Dumbbells", "Football", "Tennis Racket", "Bicycle"],
    "Toys & Games": ["LEGO Set", "Board Game", "Action Figure", "Remote Car", "Dollhouse"],
    "Automotive": ["Car Cover", "Tire Inflator", "Car Vacuum", "Dashboard Camera", "GPS Navigator"],
    "Grocery": ["Olive Oil", "Cereal", "Coffee Beans", "Pasta", "Snacks"],
    "Health & Household": ["Vitamin C", "Pain Reliever", "Toilet Paper", "Laundry Detergent", "Sanitizer"]
}

# Generate Customers
customers = pd.DataFrame([{
    "customer_id": i+1,
    "f_name": fake.first_name(),
    "l_name": fake.last_name(),
    "state": fake.state(),
    "address": fake.street_address() + ", " + fake.city() + ", " + fake.zipcode()
} for i in range(NUM_CUSTOMERS)])

# Generate Sellers
sellers = pd.DataFrame([{
    "seller_id": i+1,
    "seller_name": fake.company()
} for i in range(NUM_SELLERS)])

# Generate Categories
categories_df = pd.DataFrame([{
    "category_id": i+1,
    "category_name": category
} for i, category in enumerate(categories)])

# Generate Products
product_list = []
for i in range(NUM_PRODUCTS):
    category = random.choice(categories)
    product_list.append({
        "product_id": i+1,
        "product_name": random.choice(products[category]),
        "price": round(random.uniform(10, 1000), 2),
        "cogs": round(random.uniform(5, 500), 2),
        "category_id": categories.index(category) + 1,
        "seller_id": random.randint(1, NUM_SELLERS)
    })
products_df = pd.DataFrame(product_list)

# Generate Orders
orders = pd.DataFrame([{
    "order_id": i+1,
    "order_date": fake.date_between(start_date="-1y", end_date="today"),
    "customer_id": random.randint(1, NUM_CUSTOMERS),
    "order_status": random.choice(["Pending", "Shipped", "Delivered", "Cancelled"]),
    "product_id": random.randint(1, NUM_PRODUCTS),
    "seller_id": random.randint(1, NUM_SELLERS)
} for i in range(NUM_ORDERS)])

# Generate Order Items
order_items = pd.DataFrame([{
    "order_item_id": i+1,
    "order_id": random.randint(1, NUM_ORDERS),
    "product_id": random.randint(1, NUM_PRODUCTS),
    "quantity": random.randint(1, 5),
    "price_per_unit": round(random.uniform(10, 1000), 2),
    "total_price": round(random.uniform(10, 5000), 2)
} for i in range(NUM_ORDER_ITEMS)])

# Generate Payments
payments = pd.DataFrame([{
    "payment_id": i+1,
    "payment_date": fake.date_between(start_date="-1y", end_date="today"),
    "payment_mode": random.choice(["Credit Card", "Debit Card", "PayPal", "Cash on Delivery"]),
    "payment_status": random.choice(["Success", "Failed", "Pending"]),
    "order_id": random.randint(1, NUM_ORDERS)
} for i in range(NUM_PAYMENTS)])

# Generate Shipping
shipping = pd.DataFrame([{
    "shipping_id": i+1,
    "order_id": random.randint(1, NUM_ORDERS),
    "delivery_status": random.choice(["In Transit", "Delivered", "Returned", "Delayed"]),
    "shipping_date": fake.date_between(start_date="-1y", end_date="today"),
    "return_date": None if random.random() > 0.9 else fake.date_between(start_date="-1y", end_date="today")
} for i in range(NUM_SHIPMENTS)])

# Generate Inventory
inventory = pd.DataFrame([{
    "inventory_id": i+1,
    "product_id": random.randint(1, NUM_PRODUCTS),
    "stock_remaining": random.randint(0, 500),
    "ware_house_id": random.randint(1, 10),
    "restock_date": fake.date_between(start_date="-1y", end_date="today")
} for i in range(NUM_INVENTORY)])

# Save to CSV files
customers.to_csv("customers.csv", index=False)
sellers.to_csv("sellers.csv", index=False)
categories_df.to_csv("categories.csv", index=False)
products_df.to_csv("products.csv", index=False)
orders.to_csv("orders.csv", index=False)
order_items.to_csv("order_items.csv", index=False)
payments.to_csv("payments.csv", index=False)
shipping.to_csv("shipping.csv", index=False)
inventory.to_csv("inventory.csv", index=False)

print("✅ Amazon-like dataset generated successfully!")
```
